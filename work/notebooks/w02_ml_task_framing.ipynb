## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is one web page. Each row represents one anonymized content page with page-level information such as impressions, average position, CTR, content age, days since the last update, word count, and traffic trend. The model will assign a priority score to each page so that editors can rank individual pages for review.

```python
import os
import subprocess
import pandas as pd

repo_dir = "flyrank-ml-internship-starter"

if not os.path.isdir(repo_dir):
    subprocess.run(
        [
            "git",
            "clone",
            "--depth",
            "1",
            "https://github.com/flyrank-bih/flyrank-ml-internship-starter",
            repo_dir
        ],
        check=True
    )

data_path = os.path.join(
    repo_dir,
    "data",
    "raw",
    "content_refresh_anonymized.csv"
)

df = pd.read_csv(data_path)

page_columns = [
    "content_id",
    "impressions_90d",
    "avg_position",
    "ctr",
    "content_age_days",
    "days_since_last_update",
    "word_count",
    "trend_direction"
]

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("One row represents one web page.")

df[page_columns].head(3)
# Sketch the binary target/proxy column
df["is_declining_target"] = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .astype(int)
)

target_preview_columns = [
    "content_id",
    "impressions_90d",
    "avg_position",
    "ctr",
    "trend_direction",
    "is_declining_target"
]

print("\nTarget meaning: 1 = declining, 0 = not declining")
df[target_preview_columns].head(5)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as “review pages that are old and have high impressions” may work for some obvious cases, but it cannot easily combine several interacting signals. Refresh priority may depend on impressions, average position, CTR, content age, time since the last update, word count, and engagement at the same time. Machine learning can learn combinations and thresholds from the data and produce a continuous priority score for ranking pages. However, I would still compare the model with a simple hand-written baseline and only prefer ML if it improves the review queue on a proper client-holdout evaluation.
